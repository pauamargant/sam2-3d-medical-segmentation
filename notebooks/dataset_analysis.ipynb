```python
from pathlib import Path
import yaml
import re 
import pandas as pd
```

```python
bone_names_file = Path("../configs/dataset/bone_names.yaml")
bone_names_data = yaml.safe_load(bone_names_file.read_text())
bone_files_whitelist = set(bone_names_data['bone_files'])
dataset = Path('../data')

patients = list(dataset.glob("s*"))

raw_counts = {}
joined_counts = {}

print(f"Processing {len(patients)} patients...")

for patient in patients:
    segmentations_dir = patient / "segmentations"
    if not segmentations_dir.exists():
        continue
        
    scans = list(segmentations_dir.glob("*.nii.gz"))
    for scan in scans:
        if scan.name not in bone_files_whitelist:
            continue
            
        # 1. Count raw (without joining) - use the filename stem (or just filename without extension)
        # The prompt asks for "without joining together the ojnes with left, right, numbers"
        # I'll use the bone name logic before regex
        
        fname_str = scan.stem # .nii
        bone_name_raw = fname_str.replace(".gz", "").replace(".nii", "")
        
        if bone_name_raw not in raw_counts:
            raw_counts[bone_name_raw] = 0
        raw_counts[bone_name_raw] += 1
        
        # 2. Count joined - use the regex logic
        bone_name_joined = bone_name_raw
        
        match = re.match(r"^(.*?)_(?:left|right)_(?:[A-Z]?)(\d+)$", bone_name_joined)
        if match:
            bone_name_joined = match.group(1)
        else:
            match = re.match(r"^(.*?)_(?:left|right)$", bone_name_joined)
            if match:
                bone_name_joined = match.group(1)
            else:
                match = re.match(r"^(.*?)_(?:[A-Z]?)(\d+)$", bone_name_joined)
                if match:
                    bone_name_joined = match.group(1)
        
        if bone_name_joined not in joined_counts:
            joined_counts[bone_name_joined] = 0
        joined_counts[bone_name_joined] += 1

print("\n--- Raw Counts (Before Join) ---")
# Sort by count desc
for bone, count in sorted(raw_counts.items(), key=lambda x: x[1], reverse=True):
    print(f"{bone}: {count}")

print("\n--- Joined Counts (After Join) ---")
for bone, count in sorted(joined_counts.items(), key=lambda x: x[1], reverse=True):
    print(f"{bone}: {count}")



```
